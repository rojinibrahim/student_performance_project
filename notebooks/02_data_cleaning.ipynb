```python
import pandas as pd
import numpy as np
```

```python
student_mat = pd.read_csv("../data/raw/student-mat.csv", delimiter=";")
student_por = pd.read_csv("../data/raw/student-por.csv", delimiter=";")
student_mat_clean = student_mat.loc[student_mat['G3'] > 0].copy()
student_por_clean = student_por.loc[student_por['G3'] > 0].copy()
student_mat_clean['absences'] = student_mat_clean['absences'].clip(upper=20)
student_por_clean['absences'] = student_por_clean['absences'].clip(upper=20)
```

```python
binary_columns = ["schoolsup", "famsup", "paid", "activities", "nursery",
                  "higher", "internet", "romantic"]
for column in binary_columns:
    student_mat_clean[column] = (student_mat_clean[column] == "yes").astype(int)
    student_por_clean[column] = (student_por_clean[column] == "yes").astype(int)
```

```python
ordinal_columns = [  "Medu", "Fedu", "traveltime", "studytime","failures", "famrel",
                    "freetime", "goout", "Dalc", "Walc", "health"]
for column in ordinal_columns:
    student_mat_clean[column] = student_mat_clean[column].astype("category")
    student_por_clean[column] = student_por_clean[column].astype("category")
```

```python
student_mat_clean["pass-fail"] = (student_mat_clean["G3"] >= 10).astype(int)
student_por_clean["pass-fail"] = (student_por_clean["G3"] >= 10).astype(int)
```

```python
student_mat_clean.to_csv("../data/processed/student_mat_clean.csv", index=False)
student_por_clean.to_csv("../data/processed/student_por_clean.csv", index=False)
```

```python
print(f"Math clean shape: {student_mat_clean.shape}")
print(f"Portuguese clean shape: {student_por_clean.shape}")
print("Math target distribution:")
print(student_mat_clean["pass-fail"].value_counts())
print("Portuguese target distribution:")
print(student_por_clean["pass-fail"].value_counts())
```
