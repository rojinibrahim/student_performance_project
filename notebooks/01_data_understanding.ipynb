```python
student_mat = pd.read_csv(data\raw\"student-mat.csv", delimiter=";")
student_por = pd.read_csv(data\raw\"student-por.csv", delimiter=";")
print(f"Math shape: {student_mat.shape}")
print(f"Portuguese shape: {student_por.shape}")
print("\n--- Math dataset info ---")
student_mat.info()
print("Missing Values(mat):")
print(student_mat.isna().sum().sum())
print("Duplicated Rows(mat):")
print(student_mat.duplicated().sum())
print("\n--- Portuguese dataset info ---")
student_por.info()
print("Missing Values(por):")
print(student_por.isna().sum().sum())
print("Duplicated Rows(por):")
print(student_por.duplicated().sum())
```

```python
categorical_columns = student_mat.select_dtypes(include="object").columns.intersection(student_por.select_dtypes(include="object").columns)
for column in categorical_columns:
    print(f"--{column}--")
    mat_s = (student_mat[column].value_counts(normalize=True)*100).round(1)
    por_s = (student_por[column].value_counts(normalize=True)*100).round(1)
    comparison_df = pd.concat({"Math (%)": mat_s, "Portuguese (%)": por_s}, axis=1).fillna(0)
    print(comparison_df)
    print("\n")
```

```python
ordinal_columns = [  "Medu", "Fedu", "traveltime", "studytime","failures", "famrel",
                    "freetime", "goout", "Dalc", "Walc", "health"]
print("--- Math Ordinal Ranges ---")
for column in ordinal_columns:
    print(f"{column:<12}: {sorted(student_mat[column].unique().tolist())}")

print("\n--- Portuguese Ordinal Ranges ---")
for column in ordinal_columns:
    print(f"{column:<12}: {sorted(student_por[column].unique().tolist())}")
```
